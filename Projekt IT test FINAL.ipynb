```python
import random

#preklep jde k smrti a to je naschval
# většina věcí které dělají hru těžší jsou naschvál
# PROMĚNNÉ
inventory = []

enemies = {
    "GORTH’MAK": 80,
    "Azgorith": 100,
    "Xul’varoth": 120,
    "Ograthul": 150,
    "Baelzhar": 180,
    "Skar’morath": 200
}

health = 100
e_health = 100
patro = 0
game_over = False


# FUNKCE
def debil(text):
    global game_over

    try:
        return int(input(text))

    except ValueError:
        print(" Věž nepřijímá chyby. "
              "(Špatně jsi něco napsal nebo vůbec nic skill issue lwk)")
        game_over = True
        return None


def fight_system(boss_hp=None):
    global health
    global game_over

    enemy = random.choice(list(enemies.keys()))

    if boss_hp is None:
        e_health = enemies[enemy]
    else:
        e_health = boss_hp

    max_e_health = e_health

    print(f"Před tebou se zjevuje démon {enemy}!")

    while True:
        co_ted = debil("Zaútočit (1), použít léčení (2), nebo se vzdát (3)? ")

        if game_over:
            break

        if co_ted == 1:
            e_health -= 25
            health -= 25

            if "Blood dagger" in inventory:
                e_health -= 10

            if "Tajny mec" in inventory:
                e_health -= 20
            

        elif co_ted == 2:
            health += 35
            e_health += 20

            if health > 100:
                health = 100
                print("Už máš moc HP.")

            if e_health > max_e_health:
                e_health = max_e_health

        elif co_ted == 3:
            print("Odvrátil ses od svého osudu. Věž si tě přesto vezme.")
            game_over = True
            break

        else:
            print("Neplatná možnost.")
            continue

        print(f"Tvůj stav: {health} HP | {enemy}: {e_health} HP")

        if e_health <= 0 and e_health <= 0:
                    print("Démon padl. Cesta vzhůru pokračuje...")
                    break

        if health <= 0:
            health = 0
            print("Tvé síly opustily tvé tělo. Tvá cesta zde končí.")
            game_over = True
            break

        


# ZAČÁTEK PŘÍBĚHU
print("Probudíš se na nekonečné pláni. Kolem tebe se rozprostírá "
      "mlha a v dálce se tyčí obrovská věž, jejíž vrchol mizí v oblacích. Připadá ti podobně, ale nevíš proč.")

moznost = debil("Za tebou se ozývá děsivé vrčení. Něco tě sleduje. "
                "Vydáš se k věži (1), nebo zůstaneš na místě (2)? ")


# VSTUP DO VĚŽE
if not game_over:

    if moznost == 1:
        print("Před tebou stojí mohutná kamenná brána. Pokusíš se ji otevřít...")
        gamble = random.randint(1, 6)

        if gamble <= 2:
            print("Brána se neotevřela. Tvůj osud byl zpečetěn.(33% šance LOL)")
            game_over = True

        else:
            print("Brána se s ohlušujícím skřípěním otevírá. Vstupuješ dovnitř.")

            moznost1 = debil("Uvnitř věže potkáváš Starého muže. "
                             "Podívá se na tebe a pronese: "
                             "„Ach, synečku... Co tě přivedlo na toto prokleté místo? "
                             "Vypadáš mi nějak podobně.“ "
                             "Podává ti dýku. „Budeš ji potřebovat.“ "
                             "Poděkuješ mu (1), nebo odmítneš (2)? ")

            if not game_over:

                # PRVNÍ ROZHODNUTÍ
                if moznost1 == 1:
                    print("„Není zač,“ odpoví stařec. Poté se otočí a zmizí ve tmě.")
                    print("Pokračuješ vzhůru. Na schodech zahlédneš malé, rudě zbarvené dítě.")

                    inventory.append("dagger")
                    patro += 1

                    # PATRA 1–3
                    while patro < 4 and not game_over:
                        print(f"{patro}. patro")
                        fight_system()

                        if game_over:
                            break

                        gamble_weapon = random.randint(1, 10)

                        if gamble_weapon == 5:
                            if "Blood dagger" not in inventory:
                                inventory.append("Blood dagger")
                                print("Na zemi nacházíš Blood dagger. Jeho čepel pulzuje temnou energií. (+10 DMG)")

                        patro += 1
                        if patro == 3:
                            print("Malý démunek tě vidí a bojí se tě, vypadá to že už o tobě ví.")

                    # PATRO 4 – BOSS
                    if patro == 4 and not game_over:
                        print("PŘED TEBOU STOJÍ STRÁŽCE ČTVRTÉHO PATRA." \
                        "Když tě boss vidí říka, konečně můj praví test síly.")
                        fight_system(150)

                        if not game_over:
                            gamble_boss = random.randint(1, 10)

                            if gamble_boss in (1, 2, 3, 4):
                                print("Boss ti dá tajný meč.")
                                inventory.append("Tajny mec")

                            print("Démon padá k zemi. Jeho poslední slova znějí: Ascencion")
                            patro += 1

                    # PATRA 5–9
                    while patro < 10 and not game_over:
                        print(f"Vystoupal jsi do {patro}. patra.")
                        fight_system()

                        if game_over:
                            break
                    if patro == 7:
                        dite = debil("Vezmes dítě sebou (1) ano (2) ne")
                        if dite == 1:
                            inventory.append("Červené dítě")
                        else:
                            print("Dítě nehcáš a v dálce slyšíš jak brečí")

                        patro += 1

                    # PATRO 10 – FINÁLNÍ BOSS
                    if patro == 10 and not game_over:
                        print("GIGA ULTRA BOSS FIGHT")
                        fight_system(400)

                        if not game_over:
                            print("Zvítězil jsi. Otevíráš oči a probouzíš se ve své posteli. "
                                  "Všechno vypadá normálně... až na podivný pocit, který tě neopouští.")

                            # TAJNÉ KONCE
                            secret = input("Zkus si vzpomenout na poslední slovo. Pokud nevíš, napiš 1: ")

                            if secret == "Ascencion":
                                print("ULTRA SECRET BOSS FIGHT")
                                fight_system(1000)

                                if not game_over:
                                    print("Vzpomínky se vracejí. Démoni, které jsi porazil, jsou tvými poddanými. "
                                          "Celá tato cesta byla pouze zkouškou. Byli slabí... snad příště budou silnější." \
                                          "Ale spomínáš si, testoval jsi sám sebe a teď se nemůžes vrátit do své" \
                                          " formy nedal jsi svou vlastní skoušku. ")

                            elif secret == "dítě":
                                print("NEJTAJEMNĚJŠÍ SOUBOJ")
                                fight_system(2000)

                                if not game_over:
                                    print("Rudé dítě padlo. Však v tu chvíli začínáš mizet. "
                                          "Pravda se ti vrací do paměti. Ty jsi to dítě. "
                                          "Ty jsi skutečný král démonů.")
                                if not game_over and "dite" in inventory:
                                    print("Ty a dítě se spojíte spolu, tvoje pravá forma se vrací můžes vládnou věž zase, jako král démonu.")

                # ODMÍTNUTÍ DÝKY
                elif moznost1 == 2:
                    print("Stařec si dýku bere zpět. „Tak tedy po svém...“ "
                          "Bez zbraně nemáš šanci přežít. Tvá cesta končí.")
                    game_over = True

                else:
                    print("Neplatná možnost.")
                    game_over = True

    # ZŮSTÁNÍ NA PLÁNI
    elif moznost == 2:
        print("Z mlhy vystupuje Azgorith. Nemáš kam utéct. Jeho stín tě pohltí.")
        game_over = True

    else:
        print("Neplatná možnost.")
        game_over = True
```
